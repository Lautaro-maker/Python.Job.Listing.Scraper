```python
#Project
#Python web scraper that collects job listings from the Fake Python Jobs website.
#extract information such as the job title, company name, location, and a link to the full job description.
#Tecnologias: Python - Requests - Beautiful Soup (to parse and navigate HTML) - CSV module 
# Links: https://realpython.github.io/fake-jobs/ 
```

```python
url = "https://realpython.github.io/fake-jobs/"
```

```python
# libraries 
from bs4 import BeautifulSoup
import requests
```

```python
#download the webpage
data = requests.get(url).text
```

```python
# where find it: <div class="card-content">
    #Job Title: <h2 class="title is-5">Energy engineer</h2>
    #Company Name: <h3 class="subtitle is-6 company">Vasquez-Davidson</h3>
    #Location: <p class="location"> Christopherville, AA </p>
    #Link full job d:  <footer class="card-footer">
                            #<a href="https://realpython.github.io/fake-jobs/jobs/energy-engineer-1.html" target="_blank" class="card-footer-item">Apply</a>
                        #</footer>
```

```python
# create a soup object
soup = BeautifulSoup(data, "html.parser")
```

```python
data_div = soup.find_all('div', class_='card-content')
```

```python
for data_jobs in data_div:
    job_title = data_jobs.find('h2', class_= 'title is-5').text.strip()
    company_name = data_jobs.find('h3', class_= "subtitle is-6 company").text.strip()
    location = data_jobs.find ('p', class_= "location").text.strip()
    link_full = data_jobs.find ('a', class_= "card-footer-item", string='Apply').get('href')
    print('''
    Job Title: {}
    Company Name: {}
    Location: {}
    Link to the full job description: {} 
    '''.format(job_title, company_name, location, link_full))
    
```

```python
import csv

with open("Job.Listing.Scrap.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(["Job Title", "Company Name", "Location", "Link"])
    for data_jobs in data_div:
        job_title = data_jobs.find('h2', class_= 'title is-5').text.strip()
        company_name = data_jobs.find('h3', class_= "subtitle is-6 company").text.strip()
        location = data_jobs.find ('p', class_= "location").text.strip()
        link_full = data_jobs.find ('a', class_= "card-footer-item", string='Apply').get('href')

print("Data saved to Job.Listing.Scrap.csv")
```
